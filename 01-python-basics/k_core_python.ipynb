###creatig a  Environment in Vs Code

Way 1: Using venv (built-in Python module)
C:\Users\sivag\git-projects>python -m venv sample_env
C:\Users\sivag\git-projects>sample_env\Scripts\activate

venv comes built into Python (no install needed)
Creates a folder called sample_env with its own Python + pip
To deactivate: just type deactivate


Way 2: Using pip (via virtualenv package)
C:\Users\sivag\git-projects>pip install virtualenv
C:\Users\sivag\git-projects>virtualenv sample_env2
C:\Users\sivag\git-projects>sample_env2\Scripts\activate

virtualenv is a separate package (not built-in), installed via pip
Faster than venv and works on older Python versions too
Same activation/deactivation pattern as venv
To deactivate: deactivate


Way 3: Using conda (Anaconda/Miniconda)
C:\Users\sivag\git-projects>conda create --name sample_env3 python=3.10
C:\Users\sivag\git-projects>conda activate sample_env3

Requires Anaconda or Miniconda installed
python=3.10 lets you pick the exact Python version
Conda manages both Python packages and non-Python dependencies (useful for data science libs like numpy, scipy)
To deactivate: conda deactivate
To list all conda envs: conda env list
To delete an env: conda remove --name sample_env3 --all
