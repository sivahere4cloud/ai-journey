1.3 Logical Operator

```python
age = 28
income = 50000
credit_score = 720
 
# ALL conditions must be true
loan_approved = (age >= 21) and (income >= 30000) and (credit_score >= 700)
print(f'Loan approved: {loan_approved}') # True
 
# AT LEAST ONE condition true
needs_review = (credit_score < 600) or (income < 20000)
print(f'Needs manual review: {needs_review}') # False
```

1.5 BitWise Operator

```python
# How it works (5 in binary = 101, 3 in binary = 011)
print(5 & 3) # 101 & 011 = 001 = 1
print(5 | 3) # 101 | 011 = 111 = 7
print(5 ^ 3) # 101 ^ 011 = 110 = 6
print(5 << 1) # shift left = 1010 = 10 (doubles)
print(5 >> 1) # shift right = 10 = 2 (halves)
```

1.7 Identity Operators

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a
print(a == b) # True - same VALUES
print(a is b) # False - different OBJECTS in memory
print(a is c) # True - c points to SAME object as a
# Common use - checking for None
result = None
if result is None:
    print('No result yet') # This runs
```

Operator Precedence (Order of Execution)
Python follows this order (high to low priority):
1. () <- Parentheses (highest)
2. ** <- Exponent
3. *, /, //, % <- Multiplication / Division
4. +, - <- Addition / Subtraction
5. <, >, <=, >=, ==, != <- Comparison
6. not <- Logical NOT
7. and <- Logical AND
8. or <- Logical OR (lowest)

2.1 If Statement

```python
model_accuracy = 0.92
if model_accuracy >= 0.90:
    print('Model is excellent!')
    print('Ready for deployment')
```

2.2 if-else statement

```python
prediction_score = 0.73
if prediction_score >= 0.5:
    print('Prediction: FRAUD')
else:
    print('Prediction: LEGITIMATE')
```

2.3 if-elif-else

```python
accuracy = 0.84
if accuracy >= 0.95:
    grade = 'A - Production ready'
elif accuracy >= 0.85:
    grade = 'B - Good, minor tuning'
elif accuracy >= 0.70:
    grade = 'C - Needs improvement'
elif accuracy >= 0.50:
    grade = 'D - Poor, retrain'
else:
    grade = 'F - Failed model'
print(f'Model grade: {grade}')
# Output: Model grade: C - Needs improvement
```

3.LOOPS

3.1 for loop

```python
model_scores = [0.87, 0.91, 0.84, 0.95]
for score in model_scores:
    print(f'Model accuracy: {score * 100}%')
```

```python
# for loop with range()
for i in range(5):
    print(f'Epoch {i}')
```

```python
for i in range (1,6):
    print(f'Training Round{i}')
```

```python
for i in range(0,10,2):
    print(f'Batch {i}')
```

for loop with enumerate() - get index + value

```python
features = ['age', 'income', 'credit']
for index, feature in enumerate(features):
    print(f'Column {index}: {feature}')
```

for loop over a dictionary

```python
model_config = {'lr': 0.01, 'epochs': 100, 'batch': 32}
for key in model_config: # keys only
    print(key)
for key, value in model_config.items(): # keys + values
    print(f'{key} = {value}')
```

3.2 The while Loop

```python
accuracy = 0.60
epoch = 0
while accuracy < 0.90:
    epoch += 1
    accuracy += 0.05 # simulate improvement
print(f'Epoch {epoch}: accuracy = {round(accuracy, 2)}')
print('Target reached!')
```
