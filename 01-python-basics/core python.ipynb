Operator Precedence (Order of Execution)
Python follows this order (high to low priority):
1. () <- Parentheses (highest)
2. ** <- Exponent
3. *, /, //, % <- Multiplication / Division
4. +, - <- Addition / Subtraction
5. <, >, <=, >=, ==, != <- Comparison
6. not <- Logical NOT
7. and <- Logical AND
8. or <- Logical OR (lowest)

Set Operations (The Real Power)

```python
dataset_A = {'age', 'income', 'credit', 'loan'}
dataset_B = {'age', 'income', 'gender', 'region'}
 
# Union - all features combined
all_features = dataset_A | dataset_B
# {'age', 'income', 'credit', 'loan', 'gender', 'region'}
print(all_features)
# Intersection - features in BOTH datasets
common = dataset_A & dataset_B
# {'age', 'income'}
print(common)
# Difference - features ONLY in A
only_in_A = dataset_A - dataset_B
# {'credit', 'loan'}
print(only_in_A)
# Symmetric difference - features NOT shared
not_shared = dataset_A ^ dataset_B
# {'credit', 'loan', 'gender', 'region'}
```

TOPIC 5: Range   range( ) =>range() generates a sequence of numbers. Memory-efficient — it doesn't store all numbers, it
generates them when needed. Mostly used with loops.

Creating a Range

```python
range(5) # 0, 1, 2, 3, 4
# range(start, stop) - start to stop-1
range(2, 7) # 2, 3, 4, 5, 6
# range(start, stop, step) - with step
range(0, 10, 2) # 0, 2, 4, 6, 8
# Negative step - counting down
range(10, 0, -1) # 10, 9, 8, ..., 1
# range() shows itself, not the numbers
print(range(5)) # range(0, 5)
print(list(range(5))) # [0, 1, 2, 3, 4]
```

The L:R Rule (Important!)
L:R = Left index is included, Right index is NOT included (it's N-1)

```python
list(range(1, 6)) # [1, 2, 3, 4, 5] - 6 NOT included
list(range(0, 5)) # [0, 1, 2, 3, 4] - 5 NOT included
```

Range with Loops

```python
for epoch in range(1, 6):
    print(f'Epoch {epoch}/5')
# Epoch 1/5 ... Epoch 5/5
 
# DS/AI Example - batch processing
total_samples = 100
batch_size = 20
for batch_start in range(0, total_samples, batch_size):
    print(f'Processing batch: {batch_start} to {batch_start + batch_size}')
# 0 to 20, 20 to 40, 40 to 60, 60 to 80, 80 to 100
 
# Countdown
for i in range(5, 0, -1):
    print(i)
# 5, 4, 3, 2, 1
```
