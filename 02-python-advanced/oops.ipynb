```python
class MLModel:
    
    def __init__(self, name):
        self.name = name
        self.is_trained = False
    
    def train(self):
        print(f"Training {self.name}...")
        self.is_trained = True
    
    def predict(self):
        if not self.is_trained:
            print("Train first!")
            return
        print(f"{self.name} predicting...")


# Specific models inherit common training logic
class LinearRegression(MLModel):
    
    def calculate_slope(self):
        print(f"{self.name}: calculating slope")


class RandomForest(MLModel):
    
    def calculate_features(self):
        print(f"{self.name}: ranking features")


class NeuralNetwork(MLModel):
    
    def add_layer(self, neurons):
        print(f"{self.name}: added {neurons}-neuron layer")



m1 = LinearRegression("Sales Prediction ")
m1.train()
m1.predict()
m1.calculate_slope()

m2 = RandomForest("Feature Selection")
m2.train()
m2.predict()
m2.calculate_features()

m3 = NeuralNetwork("Image Recognition")
m3.train()
m3.predict()
m3.add_layer(128)
```

🔄 Method Overriding
A child class can rewrite a parent's method to do something different.Same method name, different implementation per class.

```python
class Animal:
    def make_sound(self):
        print("Some generic sound")


class Dog(Animal):
    def make_sound(self):              # OVERRIDES parent's class
        print("Woof! Woof!")


class Cat(Animal):
    def make_sound(self):              # OVERRIDES parent's class
        print("Meow!")


class Cow(Animal):
    def make_sound(self):              # OVERRIDES parent's class
        print("Moo!")


# Same method, different behavior per class
Animal().make_sound()  
Dog().make_sound()      
Cat().make_sound()     
Cow().make_sound()      
```

```python
# Example realistic — a vehicle system
class Vechicle:

    def __init__(self,brand,model,year):
        self.brand = brand
        self.model = model
        self.year = year
        self.engine_on = False

    def start_engine(self):
        if not self.engine_on:
            print(f"{self.brand} {self.model} engine started.")
            self.engine_on = True
        else:
            print(f"{self.brand} {self.model} engine is already on.")

    def stop_engine(self):
        if self.engine_on:
            print(f"{self.brand} {self.model} engine stopped.")
            self.engine_on = False
        else:
            print(f"{self.brand} {self.model} engine is already off.")


class Car(Vechicle):
    def __init__(self,brand,model,year,doors=4):
        super().__init__(brand,model,year)
        self.doors = doors

    def open_trunk(self):
        print(f"{self.brand} {self.model} trunk is open.")

class Motorcycle(Vechicle):
    def __init__(self,brand,model,year,type="Sport"):
        super().__init__(brand,model,year)
        self.type = type

    def pop_wheelie(self):
        if self.engine_on:
            print(f"{self.brand} {self.model} is popping a wheelie!")
        else:
            print(f"Start the engine first to pop a wheelie!")


class ElectricCar(Car):
    def __init__(self,brand,model,year,doors=4,battery_capacity=75):
        super().__init__(brand,model,year,doors)
        self.battery_capacity = battery_capacity

    def charge(self):
        print(f"{self.brand} {self.model} is charging... Battery at {self.battery_capacity}%")




Car =Car("Tesla", "Model S", 2022)
Car.start_engine()
Car.open_trunk()

Motorcycle = Motorcycle("Yamaha", "YZF-R3", 2021)
Motorcycle.start_engine()
Motorcycle.pop_wheelie()

ElectricCar = ElectricCar("Nissan", "Leaf", 2020)
ElectricCar.start_engine()
ElectricCar.charge()    
















```

```python
# Parent class
class Employee:
    
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    
    def display(self):
        print(f"{self.name} earns ${self.salary}")
    
    def annual_salary(self):
        return self.salary * 12


# Child class — Manager
class Manager(Employee):
    
    def __init__(self, name, salary, team_size):
        super().__init__(name, salary)
        self.team_size = team_size
    
    def manage(self):
        print(f"{self.name} is managing {self.team_size} people")


# Child class — Developer
class Developer(Employee):
    
    def __init__(self, name, salary, language):
        super().__init__(name, salary)
        self.language = language
    
    def code(self):
        print(f"{self.name} is coding in {self.language}")


# Test
m = Manager("Sarah", 8000, 10)
m.display()              # inherited
m.manage()               # own method
print(m.annual_salary()) # inherited

d = Developer("Ravi", 5000, "Python")
d.display()              # inherited
d.code()                 # own method
print(d.annual_salary()) # inherited
```

```python
class LinearRegression:
    def predict(self, data):
        return "Predicting using straight-line formula"

class RandomForest:
    def predict(self, data):
        return "Predicting using tree-based ensemble"

class NeuralNetwork:
    def predict(self, data):
        return "Predicting using deep layers"

class XGBoost:
    def predict(self, data):
        return "Predicting using gradient boosting"


# ONE function works for ANY model — that's polymorphism!
def evaluate_model(model, test_data):
    result = model.predict(test_data)
    print(result)

# Use any model interchangeably
models = [LinearRegression(), RandomForest(), NeuralNetwork(), XGBoost()]
test_data = [1, 2, 3]

for model in models:
    evaluate_model(model, test_data)
```

```python
 #Two Types of Polymorphism
#Type 1: Method Overriding (Inheritance-Based) — From Last Lesson
#When a child class rewrites a method from its parent:

class Animal:
    def speak(self):
        print("Some generic sound")

class Dog(Animal):
    def speak(self):              # OVERRIDES parent's
        print("Woof!")

class Cat(Animal):
    def speak(self):              # OVERRIDES parent's
        print("Meow!")
```

```python
#Type 2: Duck Typing (Python's Special Style) 
#This is uniquely Python. Here's the philosophy:

#"If it walks like a duck and quacks like a duck, it's a duck."

#In Python, classes don't even need to be related to be polymorphic. They just need to have the same method name.

class Dog:
    def speak(self):
        return "Woof!"

class Robot:
    def speak(self):
        return "Beep boop!"

class Person:
    def speak(self):
        return "Hello!"

# These classes are TOTALLY UNRELATED
# But our code treats them the same way

things = [Dog(), Robot(), Person()]

for thing in things:
    print(thing.speak())
```

```python
#Polymorphism with Inheritance (The Full Picture)
class Shape:
    def __init__(self, name):
        self.name = name
    
    def area(self):
        print("Calculating area...")     # generic version
    
    def describe(self):
        print(f"This is a shape: {self.name}")


# Child class — overrides area()
class Circle(Shape):
    def __init__(self, radius):
        super().__init__("Circle")
        self.radius = radius
    
    def area(self):                       # POLYMORPHIC override
        return 3.14 * self.radius ** 2


# Child class — overrides area()
class Square(Shape):
    def __init__(self, side):
        super().__init__("Square")
        self.side = side
    
    def area(self):                       # POLYMORPHIC override
        return self.side ** 2


# Child class — overrides area()
class Triangle(Shape):
    def __init__(self, base, height):
        super().__init__("Triangle")
        self.base = base
        self.height = height
    
    def area(self):                       # POLYMORPHIC override
        return 0.5 * self.base * self.height


# Polymorphism magic — same code, different shapes
shapes = [Circle(5), Square(4), Triangle(3, 6)]

for shape in shapes:
    shape.describe()                      # inherited
    print(f"Area: {shape.area()}")        # polymorphic — different per class
    print()
```

```python
class CreditCard:
    def __init__(self, holder, amount):
        self.holder = holder
        self.amount = amount
    
    def process(self):
        print(f"Processing ${self.amount} via Credit Card for {self.holder}")


class PayPal:
    def __init__(self, holder, amount):
        self.holder = holder
        self.amount = amount
    
    def process(self):
        print(f"Processing ${self.amount} via PayPal for {self.holder}")


class Crypto:
    def __init__(self, holder, amount):
        self.holder = holder
        self.amount = amount
    
    def process(self):
        print(f"Processing ${self.amount} via Crypto for {self.holder}")


# Polymorphism — one function handles ANY payment method
def checkout(payment):
    payment.process()


# Test
checkout(CreditCard("Ravi", 100))      # via Credit Card
checkout(PayPal("Sita", 50))           # via PayPal
checkout(Crypto("Anil", 200))          # via Crypto

# Add a new payment method? Just create a class with process() method!
```
