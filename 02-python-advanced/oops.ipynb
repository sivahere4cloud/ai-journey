OOP = Object-Oriented Programming
 It's a way of writing code that organizes data and functions together into "objects" — just like things in the real world.
 Everything in the real world has:

Properties (data about it)
Actions (things it can do)
Class = Blueprint
Object = Actual Thing


```python
class Dog:
    pass

buddy =Dog()
max = Dog()
rocky = Dog()
print(buddy)
print(max)  
print(rocky)
print(type(buddy))
print(type(max))
print(type(rocky))
```

```python
class Dog:
    pass

buddy = Dog()
buddy.name = "Buddy"
buddy.age = 3
buddy.breed = "Golden Retriever"


max = Dog()
max.name = "Max"
max.age = 5
max.breed = "Labrador Retriever"

rocky = Dog()
rocky.name = "Rocky"
rocky.age = 2
rocky.breed = "Bulldog"

print(buddy.name)
print(max.name)     
print(rocky.name)
```

```python
#Adding Methods (Actions) to a Class

class Dog:
    def bark(self):
        print("Woof!")
    def sleep(self):
        print("Zzz...")

Buddy = Dog()
Buddy.bark()
Buddy.sleep()

max = Dog()
max.bark()
max.sleep()

```

```python
class Dog:
    def introduce(self):
        print(f"I am a dog.{self.name} and I am {self.age} years old.")

buddy = Dog()
buddy.name = "Buddy"
buddy.age = 3

max = Dog()
max.name = "Max"
max.age = 5

buddy.introduce()
max.introduce()
```

```python
#example

class Student:
    def study(self):
        print(f"{self.name} is studying {self.subject}.")
    
    def take_brake(self):
        print(f"{self.name} is taking a break.")




ravi= Student()
ravi.name = "Ravi"
ravi.subject = "Math"

sita    = Student()
sita.name = "Sita"
sita.subject = "Science"

ravi.study()
sita.study()
ravi.take_brake()
sita.take_brake()
```

```python
#Example2

class Car:

    def start(self):
        print(f"{self.brand} {self.model} is starting...")


    def stop(self):
        print(f"{self.brand} {self.model} is stopping...")



bmw = Car()
bmw.brand = "BMW"
bmw.model = "X5"

audi = Car()
audi.brand = "Audi"
audi.model = "A4"

bmw.start()
audi.start()
bmw.stop()
audi.stop()
```
