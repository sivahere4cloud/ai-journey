OOP = Object-Oriented Programming
 It's a way of writing code that organizes data and functions together into "objects" — just like things in the real world.
 Everything in the real world has:

Properties (data about it)
Actions (things it can do)
Class = Blueprint
Object = Actual Thing

```python
class Student:
    def __init__(self, name, age, subject, marks):
        self.name = name
        self.age = age
        self.subject = subject
        self.marks = marks

    def display_info(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Subject: {self.subject}")
        print(f"Marks: {self.marks}")

    def is_passed(self):
        if self.marks >= 50:
            return True
        else:
            return False
        
student1 = Student("Ravi", 20, "Math", 85)
student2 = Student("Sita", 22, "Science", 45)
student3 = Student("Gita", 21, "History", 75)

student1.display_info()
print(f"Passed: {student1.is_passed()}")
```

```python
#__init__ is used in real Data Science:

class MLModel:

    def __init__(self,name,algorithm,learning_rate=0.01,epochs=100):
        self.name = name
        self.algorithm = algorithm
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.trained = False
        self.accuracy = None
    def train(self):
        print(f"Training {self.name}  for  {self.epochs} epochs...")
        self.trained = True
        self.accuracy = 0.94

    def predict(self, data):
        if not self.trained:
            print("Model is not trained yet!")
            return None
        print(f"Predicting with {self.name} predicinting at {self.accuracy*100}% accuracy...   ")

model_a = MLModel("Bank Risk Model","XGboost")
model_b = MLModel("Fraud Detect","Random Forest",0.001,500)

model_a.train()
model_a.predict([1, 2, 3])

model_b.train()
model_b.predict([4, 5, 6])


    
```

```python
class Phone:
    
    def __init__(self, brand, model, price, color="Black"):
        self.brand = brand
        self.model = model
        self.price = price
        self.color = color
    
    def display_info(self):
        print(f"{self.brand} {self.model}")
        print(f"Color: {self.color}")
        print(f"Price: ${self.price}")
    
    def is_premium(self):
        if self.price >= 1000:
            print(f"{self.brand} {self.model} is a premium phone")
        else:
            print(f"{self.brand} {self.model} is a regular phone")

# Test it
phone1 = Phone("Apple", "iPhone 15", 1200)
phone2 = Phone("Samsung", "Galaxy A14", 300, "Blue")
phone3 = Phone("Google", "Pixel 8", 900, "White")

phone1.display_info()
phone1.is_premium()

phone2.display_info()
phone2.is_premium()
```
