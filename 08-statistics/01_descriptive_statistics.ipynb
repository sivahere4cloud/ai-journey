```python
import numpy as np
import pandas as pd
from scipy import stats

data = np.array([25, 35, 28, 45, 30, 40, 22, 38])


print("=" * 40)
print("CENTRAL TENDENCY")
print("=" * 40)
print(f"Mean:     {np.mean(data):.2f}")
print(f"Median:   {np.median(data):.2f}")
print(f"Mode:     {stats.mode(data).mode}")



```

```python
print()
print("=" * 40)
print("VARIABILITY")
print("=" * 40)
print(f"Range:    {np.max(data) - np.min(data)}")
print(f"Variance: {np.var(data, ddof=1):.2f}")
print(f"Std Dev:  {np.std(data, ddof=1):.2f}")
Q1  = np.percentile(data, 25)
Q3  = np.percentile(data, 75)
IQR = Q3 - Q1
print(f"Q1:       {Q1}")
print(f"Q3:       {Q3}")
print(f"IQR:      {IQR}")
```

```python
print()
print("=" * 40)
print("ASYMMETRY")
print("=" * 40)
print(f"Skewness: {stats.skew(data):.2f}")
print(f"Kurtosis: {stats.kurtosis(data):.2f}")
```

```python
print()
print("=" * 40)
print("OUTLIERS (IQR Method)")
print("=" * 40)
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = data[(data < lower) | (data > upper)]
print(f"Lower fence: {lower}")
print(f"Upper fence: {upper}")
print(f"Outliers:    {outliers}")
```

```python
import numpy as np
import pandas as pd
from scipy import stats

def full_stats_report(data, column_name):
    """
    Complete statistical analysis of any column.
    Use this before training any ML model.
    """
    print(f"\n{'='*45}")
    print(f"  STATS REPORT: {column_name}")
    print(f"{'='*45}")

    # Central Tendency
    print("\n CENTRAL TENDENCY")
    print(f"  Mean:     {np.mean(data):.2f}")
    print(f"  Median:   {np.median(data):.2f}")
    print(f"  Mode:     {stats.mode(data).mode}")

    # Variability
    print("\n VARIABILITY")
    print(f"  Std Dev:  {np.std(data, ddof=1):.2f}")
    print(f"  Variance: {np.var(data, ddof=1):.2f}")
    Q1  = np.percentile(data, 25)
    Q3  = np.percentile(data, 75)
    IQR = Q3 - Q1
    print(f"  Q1:       {Q1:.2f}")
    print(f"  Q3:       {Q3:.2f}")
    print(f"  IQR:      {IQR:.2f}")
    print(f"  Min:      {np.min(data):.2f}")
    print(f"  Max:      {np.max(data):.2f}")
    print(f"  Range:    {np.max(data) - np.min(data):.2f}")

    # Asymmetry
    print("\n ASYMMETRY")
    skew = stats.skew(data)
    kurt = stats.kurtosis(data)
    print(f"  Skewness: {skew:.2f}", end="  ")
    if   abs(skew) <= 0.5: print(" Symmetric")
    elif abs(skew) <= 1.0: print("  Moderate skew")
    else:                  print(" High skew → fix needed")

    print(f"  Kurtosis: {kurt:.2f}", end="  ")
    if   abs(kurt) <= 1:   print(" Normal tails")
    elif kurt > 1:         print("  Heavy tails → check outliers")
    else:                  print("  Light tails")

    # Outliers
    print("\n OUTLIERS (IQR Method)")
    lower   = Q1 - 1.5 * IQR
    upper   = Q3 + 1.5 * IQR
    out     = data[(data < lower) | (data > upper)]
    print(f"  Lower fence: {lower:.2f}")
    print(f"  Upper fence: {upper:.2f}")
    if len(out) == 0:
        print("  Outliers:    None found")
    else:
        print(f"  Outliers:    {out}")


# USE IT ON ANY DATA
income = np.array([25000,30000,35000,40000,45000,50000,500000])
full_stats_report(income, "Income")
```
